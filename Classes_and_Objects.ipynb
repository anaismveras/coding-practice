```python
# creating the Employee class
class Employee:
  # every class has an init function - special function that gets invoted
  # when creating an object of the class
  # self - reference to the object
  def __init__(self, name, position, salary):
    self.name = name
    self.position = position
    self.salary = salary

  # operation for the class
  def display_info(self):
    return f"{self.name} is a {self.position} earning {self.salary} annually"
```

```python
# creating objects for this class and passing the data to the class
employee1 = Employee("John Doe", "Manager", 75000)
employee2 = Employee("Jane Smith", "Developer", 60000)
```

```python
# printing the classes that were just created
print(employee1.display_info())
print(employee2.display_info())
```

```python
# change the attributes of an object
employee1.salary = 80000
print(employee1.display_info())
```
