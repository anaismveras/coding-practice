```python
import pandas as pd

df = pd.read_csv('/content/Order.csv')
```

```python
new_df = df.dropna() #will drop any rows that have a cell of null value
print(new_df.to_string())
```

```python
df.fillna(100, inplace=True)
#fill in any null columns with a default value, inplace=true - dont give a new dataframe just do the change in this dataframe
```

```python
df["UnitsSold"].fillna(100, inplace=True)
#all of the values in this specfific column will be replaced with value given
```

```python
df_new = df.drop_duplicates() # will remove all the duplicate rows int eh dataframe
```

```python
# # function where the date format might not be correct
# df["OrderDate"] = pd.to_datetime(df["OrderDate"], format='%d/%m/%Y', errors='ignore')
# errors will try to format the date to the format if the exisiting value can not be formated it will not cause errors, it will relace erros with None, Null value
```

# Data Manupulation and Aggregation

```python
avg_sale = new_df.groupby('ItemType')['UnitsSold'].mean()
print(avg_sale)
# the average sale for each product catogery
#doing a groupby on new_df by ItemType and for each of the ItemTypes we are getting
```

```python
new_df.aggregate(['sum', 'min'])
```

```python
# merge new file with exisitng one
new_order_file = pd.read_csv('/content/Order-New-New.csv')
# merge the data on one specific column
merge_data = pd.merge(new_df, new_order_file, on='OrderID')
print(merge_data)
```

# Data Visualization

```python
import pandas as pd
import matplotlib.pyplot as plt

data = { 'Date': ['2023-08-01', '2023-08-02', '2023-08-03'],
        'Price': [100, 105, 110]}

df = pd.DataFrame(data)
df['Date'] = pd.to_datetime(df['Date'])
df.plot(x='Date', y='Price', kind='line')
plt.xlabel('Date')
plt.ylabel('Price')

plt.show()
```

```python
data = {
    'Product': ['A', 'B', 'C'],
    'Sales': [100, 150, 200]
}
df = pd.DataFrame(data)
df.plot(x='Product', y='Sales', kind='bar')
plt.xlabel('Product')
plt.ylabel('Sales')
plt.show()
```

```python
data = {'Hours_Studied': [2,4,5,6,7,8,10],
        'Exam_Score': [60, 70, 80, 75, 90, 85, 95]}
df = pd.DataFrame(data)
df.plot(x='Hours_Studied', y='Exam_Score', kind='scatter')
plt.show()
```

```python
data = {"Toppings": ['Cheese', 'Mushroom', 'Pepperoni', 'Olvies', 'Sausage', 'Pineapples'],
        'Count': [25, 40, 10, 5, 100, 20]}
df = pd.DataFrame(data)
df.plot(x='Toppings', y='Count', kind='pie')
plt.show()
```
